```python
from pathlib import Path
from PIL import Image
from ultralytics import YOLO

# Load a pretrained YOLO26n model
model = YOLO("yolo26n.pt")
jpg_files = list(Path("/content/examples").rglob("*.png"))

for file in jpg_files:
    img = Image.open(file)
    print(file)
```

```python
for file in jpg_files:
  tmp = cv2.imread(file)
  cv2_imshow(tmp)
  results = model.predict(source=tmp, save=True)
  tmp2 = cv2.imread('/content/runs/detect/predict/image0.jpg')
  cv2_imshow(tmp2)
```
